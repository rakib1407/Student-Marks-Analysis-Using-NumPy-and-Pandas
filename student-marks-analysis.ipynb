```python
import numpy as np 
import pandas as pd
```

Dataset Load

```python
df=pd.read_csv("/kaggle/input/datasets/spscientist/students-performance-in-exams/StudentsPerformance.csv")

print(df)
```

Total And  Average Marks

```python
# Total marks column
df['Total'] = df[['math score', 'reading score', 'writing score']].sum(axis=1)

# Average marks column
df['Average'] = df[['math score', 'reading score', 'writing score']].mean(axis=1)

print(df[['math score','reading score','writing score','Total','Average']].head(1000))
```

```python
print(df[['Average', 'Result']].head())
```

```python
df['Name'] = ['Student ' + str(i) for i in range(len(df))]

print(df[['Name', 'Average', 'Result']].head())
```

```python
print(df.columns)
```

**Pass/Fail Column**

```python
df['Result'] = np.where(df['Average']>=50, 'Pass', 'Fail')
print(df[['Name','Average','Result']].head(1000)) 
```

**total pass & fail**

```python
total_pass = (df['Result'] == 'Pass').sum()
total_fail = (df['Result'] == 'Fail').sum()

print("Total Pass:", total_pass)
print("Total Fail:", total_fail)
```

**Top & Bottom Scorer**

```python
top_student = df.loc[df['Total'].idxmax()]
bottom_student = df.loc[df['Total'].idxmin()]

print("Highest Scorer:\n", top_student)
print("\nLowest Scorer:\n", bottom_student)
```

Subject-wise Average Marks

```python
subject_avg = df[['math score', 'reading score', 'writing score']].mean()
print("\nSubject-wise Average:\n", subject_avg)
```

**Gender-wise Analysis**

```python
gender_avg = df.groupby('gender')[['math score', 'reading score', 'writing score']].mean()
gender_pass_rate = df.groupby('gender')['Result'].apply(lambda x: (x=='Pass').mean()*100)

print("\nGender-wise Average Marks:\n", gender_avg)
print("\nGender-wise Pass Rate (%):\n", gender_pass_rate)
```

**Test Preparation Course Analysis**

```python
prep_avg = df.groupby('test preparation course')[['math score','reading score','writing score']].mean()
prep_pass_rate = df.groupby('test preparation course')['Result'].apply(lambda x: (x=='Pass').mean()*100)

print("\nPreparation Course Average Marks:\n", prep_avg)
print("\nPreparation Course Pass Rate (%):\n", prep_pass_rate)
```

**Subject Correlation**

```python
correlation = df[['math score','reading score','writing score']].corr()
print("\nSubject-wise Correlation:\n", correlation)
```
